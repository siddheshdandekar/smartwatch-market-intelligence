### Data Type Conversion.

```python
# Convert 'price' to numeric (Float)
# errors='coerce' turns any non-numeric data into NaN safely
df['price'] = pd.to_numeric(df['price'], errors='coerce')

# Ensure 'rating' is a float
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
```

#### Numeric Conversion: Handling 'K' and 'L' in Reviews Count

```python
def parse_reviews(val):
    if pd.isna(val):
        return val
    
    val = str(val).lower().strip()
    
    if 'k' in val:
        return float(val.replace('k', '')) * 1000
    elif 'l' in val:
        return float(val.replace('l', '')) * 100000
    
    return val
```

```python
# Standardize shorthand values
df['reviews_count'] = df['reviews_count'].apply(parse_reviews)

# Convert the entire column to numeric float
df['reviews_count'] = pd.to_numeric(df['reviews_count'], errors='coerce')
df.head(3)
```

### Data Standardization

```python
# Standardize the 'title' column
# We lowercase everything and strip extra spaces
df['title'] = df['title'].str.lower().str.strip()

df.head(3)
```

### Remove duplicate rows (identical titles, prices, and ratings)

```python
df.drop_duplicates(inplace=True)
df.shape
```

### Applying a Minimum Price Threshold

```python
df = df[df['price'] >= 800]
```

### Why We Filter Products Below ₹800
Our project aims to predict the popularity of smartwatches. During data acquisition, it's common for search results to include related accessories like watch bands, screen protectors, or basic fitness trackers that are not true smartwatches.

Products priced below ₹800 are highly likely to fall into this "accessory" category. By filtering out these low-priced items, we ensure that:

* Our dataset is focused solely on actual smartwatches, improving the relevance of our analysis.

* Our machine learning model learns from meaningful data, preventing it from being misled by the popularity of non-smartwatch items.

This step helps to define the scope of what we consider a "smartwatch" for the purpose of our popularity prediction.

### Generating Feature Columns from Product Titles

```python
# Define a highly specific list of keywords to look for
specs = [
    "amoled", "hd display", "lcd", "oled", # Display Technology
    
   
    "gps", "bluetooth", "calling", "alexa",  # Connectivity & Features
    
   
    "heart rate", "spo2", "sleep", "blood oxygen",
    "sports", "fitness tracker", "step counter",  # Health & Fitness Tracking
    
    
    "waterproof", "ip68", "ip67", "5atm",# Durability / Water Resistance
    
    "men", "women", "kids" # Target Audience
]

# 3. Iterate through the list and create a new column for each keyword
for spec in specs:
    # Clean the column name (e.g., 'heart rate' becomes 'has_heart_rate')
    col_name = f'{spec.replace(" ", "_")}'
    
    # Convert title to string (to avoid errors on empty rows), make lowercase, and check if keyword exists
    df[col_name] = df['title'].astype(str).str.lower().apply(lambda x: 1 if spec in x else 0)

# 4. Preview the newly created columns
feature_cols = [f'{spec.replace(" ", "_")}' for spec in specs]

# Display the title alongside some of the newly extracted boolean columns
display(df[['title', 'amoled', 'hd_display', 'gps', 'heart_rate']].head())
```

Result: Successfully transformed unstructured product titles into 22 structured boolean features (0 or 1). This step makes the text data machine-readable, allowing us to analyze which specific technologies (like AMOLED or GPS) drive a smartwatch's price and market popularity in our upcoming Exploratory Data Analysis and ML modeling phases.

### Market Segmentation

# MACHINE LEARNING MODEL

### Outlier Treatment (Price)
* The price variable in our dataset exhibits a heavy right-skew, with values ranging from ₹1,500 to over ₹80,000. Including high-premium flagship devices (e.g., Apple Watch, Garmin) creates a significant distribution gap, as these products follow a different demand-and-supply logic compared to the mass-market listings. By applying the IQR Method, we statistically identify devices priced above the upper bound . Removing these outliers allows the model to achieve better convergence and prevents the feature weights from being disproportionately influenced by a small percentage of high-value observations, thereby improving the model's reliability for the core market segment.

```python
# Calculate using the IQR Method
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

# The standard formula is Q3 + 1.5 * IQR
upper_bound = Q3 + (1.5 * IQR)

print(f"Statistical Upper Bound for Price: ₹{upper_bound:.2f}")

# Filter the dataset to remove the outliers
df_clean = df[df['price'] <= upper_bound].copy()

# 3. Compare the sizes
rows_removed = len(df) - len(df_clean)
print(f"Original Dataset Size: {len(df)} rows")
print(f"Cleaned Dataset Size:  {len(df_clean)} rows")
print(f"Total Outliers Removed: {rows_removed} (Approximately {100*rows_removed/len(df):.1f}%)")
```

**Result :** The outlier treatment using the IQR method resulted in the removal of 101 data points, representing approximately 5.8% of the original dataset. By setting the upper price boundary at ₹16,245.88, we have effectively filtered out extreme high-end luxury devices that deviate from standard consumer purchasing patterns. This step ensures the dataset is now statistically more uniform, allowing our machine learning model to focus on the high-volume, competitive smartwatch market where feature sets and price play the most significant role in determining product popularity.

### Defining the Target Variable (Popular)
* To establish a high standard for market success, we define our target variable using the 80th percentile of reviews_count. By setting the threshold at this level, we filter out moderately successful products and focus the model on the "market leaders"—products that have achieved significant consumer engagement. A product is labeled "Popular" only if it exceeds  80th percentile of reviews_count while maintaining a high customer satisfaction rating of 4.0 or more. This rigorous definition ensures that the model learns the specific technical signatures required to reach the top tier of the smartwatch market, rather than just average performance.

```python
# Determine the 70th percentile for review counts
review_threshold = df_clean['reviews_count'].quantile(0.80)

# Generate the 'is_popular' binary column
# Criteria: Reviews >= 70th percentile AND Rating >= 4.0
df_clean['is_popular'] = ((df_clean['reviews_count'] >= review_threshold) & (df_clean['rating'] >= 4.0)).astype(int)

# Analyze the distribution of our new classes
print(f"Popularity Threshold: {review_threshold:.0f} reviews")
print("\nTarget Variable Distribution (is_popular):")
print(df_clean['is_popular'].value_counts())
```

**Result:** With the popularity threshold set at the 80th percentile, the dataset is now labeled with a binary target. 231 of the watches are classified as Popular, while 1,418 are classified as Non-Popular. This creates a significant "class imbalance". However, because these 231 products represent the most successful items in the dataset, the "signal" for popularity is much stronger, providing a clearer target for the Random Forest algorithm to identify high-impact features.

### Feature Selection & Data Splitting
* In this step, we select the independent variables (X) and the dependent variable (y). We are specifically using technical specifications like AMOLED, GPS, and Bluetooth Calling as features to determine if hardware alone can predict market success. To evaluate the model fairly, we perform a Stratified Train-Test Split with an 80/20 ratio. Stratification is vital because "Popular" products are the minority class; this technique ensures that both the training set and the testing set have the same proportion of popular watches, preventing the model from becoming biased or failing to learn the characteristics of successful products.

```python
# Select the technical features (What the model 'sees')
features = ['amoled', 'hd_display', 'gps', 'calling', 'heart_rate', 
            'spo2', 'waterproof', 'men', 'women']

X = df_clean[features]
y = df_clean['is_popular']

# 2. Split the data into Training (80%) and Testing (20%)
# stratify=y is the key to handling the class imbalance correctly
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=47, stratify=y
)

# 3. Verify the split
print(f"Training set size: {X_train.shape[0]} samples")
print(f"Testing set size:  {X_test.shape[0]} samples")
print("\nCheck if Popularity ratio is maintained (Stratification):")
print(f"Training % Popular: {y_train.mean():.2%}")
print(f"Testing % Popular:  {y_test.mean():.2%}")
```

**Result:** The dataset was successfully partitioned into 1,319 training samples and 330 testing samples. The stratification process successfully maintained the class balance, with 14.03% popular watches in the training set and 13.94% in the testing set. This consistency ensures that the model is trained and evaluated on representative samples of the market, preventing bias and ensuring the final accuracy metrics are reliable.

### Hyperparameter Tuning (GridSearchCV)
* In machine learning, a model's default settings are not always ideal for every dataset. We use GridSearchCV to automate the search for the best hyperparameters. We specifically test different forest sizes (n_estimators) and tree depths (max_depth) to find the balance between learning the data and avoiding overfitting. Because our target variable is imbalanced (only ~14% of watches are labeled as "Popular"), we initialize the model with class_weight='balanced'. This ensures the algorithm assigns higher importance to the minority class, preventing it from prioritizing majority class accuracy over minority class recall.

```python
# Define the Parameter Grid 
# We expand the grid to find the most robust settings for our smartwatch data
param_grid = {
    'n_estimators': [200, 400, 600],          
    'max_depth': [15, 25, None],                
    'min_samples_split': [2, 5, 10],                
    'criterion': ['entropy', 'gini']
}

# 2. Initialize the Random Forest with Balanced Weights
# This is our strategy to handle the class imbalance identified in Step 2
rf_clf = RandomForestClassifier(random_state=99, class_weight='balanced')

# 3. Setup GridSearchCV with 10-fold Cross-Validation
# Scoring is set to 'accuracy' to align with the reference methodology
gs = GridSearchCV(estimator=rf_clf, param_grid=param_grid, scoring='accuracy', cv=10, n_jobs=-1)

# 4. Fit the model on the Training Data
gs.fit(X_train, y_train)

# 5. Extract the best performing model
best_model = gs.best_estimator_

# 6. Display the winning parameters
print(f"Best Parameters Found: {gs.best_params_}")
```

**Result:** The GridSearchCV successfully identified that a Gini-based forest with a maximum depth of 15 and 600 trees provided the highest predictive stability. By testing across 10 separate folds, we confirmed that these parameters are not only optimized for the training data but also maintain consistent performance across different segments of the smartwatch market.

### Evaluation metrics
* We evaluate the 600-tree optimized Random Forest on the reserved testing set. Given that our "Popular" class is a minority (the top 14% of the market), evaluating the model based on Accuracy alone could be misleading. Therefore, we utilize the Classification Report to examine Precision, Recall, and the F1-Score. We specifically prioritize Recall for the Popular class, as this metric determines the model's ability to successfully "capture" high-performing products out of the total population. Our goal is to verify if the technical specifications provided to the model contain a strong enough signal to reliably distinguish market leaders from average products.

```python
# Generate final predictions on the unseen test set
y_pred = best_model.predict(X_test)

# Calculate and display the results
accuracy = accuracy_score(y_test, y_pred)

print(f"Final Optimized Model Accuracy: {accuracy:.2f}")

print("\nDetailed Classification Report:")
print(classification_report(y_test, y_pred))
```

### Feature Importance Analysis
* To provide actionable insights, we extract the Feature Importance from our optimized 600-tree Random Forest ensemble. While the previous metrics confirmed that the model can predict popularity, this analysis explains how it does so. By calculating the "Gini Importance" (Mean Decrease in Impurity), we can rank each technical specification based on its contribution to the model's decision-making process. This allows us to identify the "Key Value Drivers"—the specific hardware features that the Indian market prioritizes most when selecting a top-tier smartwatch.

```python
# 1. Extract feature importance scores
importances = best_model.feature_importances_
feat_df = pd.DataFrame({'Specification': features, 'Importance': importances})
feat_df = feat_df.sort_values(by='Importance', ascending=False)

# 2. Visualize the drivers of success
plt.figure(figsize=(10, 6))
sns.barplot(x='Importance', y='Specification',hue= 'Specification', data=feat_df, palette='viridis')

plt.title('Top Technical Drivers of Smartwatch Popularity', fontsize=14)
plt.xlabel('Relative Importance Score', fontsize=12)
plt.ylabel('Hardware Feature', fontsize=12)
plt.show()

# 3. Display the ranking
print("Final Technical Priority Ranking:")
display(feat_df)
```

**Result:** The Feature Importance analysis reveals a highly competitive "feature cluster" at the top of the ranking. Calling (14.6%), Waterproofing (14.5%), and SpO2 monitoring (14.0%) emerged as the three most critical drivers of market success. The narrow margin between these top features suggests that popularity is not driven by a single technical feature but rather by a baseline requirement for versatility. Combining communication, durability, and health tracking. Interestingly, AMOLED (13.6%) and Heart Rate monitoring (13.1%) also hold significant weight, while GPS (6.8%) ranks lower, suggesting that for the mass market represented in this dataset, daily lifestyle connectivity is a higher priority than specialized outdoor navigation.
